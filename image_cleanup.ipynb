````python
from pathlib import Path
from PIL import Image

base = Path(r'd:\tugas\coffee-shop-garage 18\image')
paths = [
    base / '1.jpeg',
    base / '3.jpeg',
    base / 'kopi gula aren.png',
    base / 'kopi-gula-aren-slider.png',
]

for src in paths:
    img = Image.open(src).convert('RGBA')
    px = img.load()
    w, h = img.size
    removed = 0
    for y in range(h):
        for x in range(w):
            r, g, b, a = px[x, y]
            if r < 40 and g < 40 and b < 40:
                px[x, y] = (0, 0, 0, 0)
                removed += 1
    out = src.with_name(src.stem + '-transparent.png')
    img.save(out, format='PNG')
    print(f'{src.name} -> {out.name} (pixels removed: {removed})')
````
